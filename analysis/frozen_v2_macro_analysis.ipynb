This notebook is a thin, reproducible wrapper around `scripts/build_frozen_analysis_pack.py`. The script writes all tables and SVG figures used below.

```python
from pathlib import Path
import json
import subprocess
import sys

try:
    from IPython.display import display as _display
except Exception:
    _display = None

def show(obj):
    if _display is not None:
        _display(obj)
    else:
        print(obj.to_string(index=False) if hasattr(obj, 'to_string') else obj)

ROOT = Path.cwd()
ATTESTATIONS = ROOT / 'frozen_data_v2' / 'attestations_frozen.csv'
TARGET = ROOT / 'frozen_data_v2' / 'target_binding_analysis_frozen.csv'
OUT_DIR = ROOT / 'analysis' / 'frozen_v2'
FIG_DIR = ROOT / 'reports' / 'figures' / 'frozen_v2'
TABLE_DIR = ROOT / 'reports' / 'tables' / 'frozen_v2'

print(ATTESTATIONS.exists(), ATTESTATIONS)
print(TARGET.exists(), TARGET)
```

## 9. When the Trend Is Too Obvious: Exceptions and Robustness

The overall binding signal is saturated. That means the analytic target shifts from `is there a trend?` to four sharper questions:

1. Which period/corridor/source combinations remain weakly bound?
2. Which source types inflate or suppress the overall binding index?
3. Do target weak-control records cluster in specific corridors?
4. Which concrete records can be used as close-reading counterexamples?

## 10. Recommended Thesis Evidence Chain

1. Corpus engineering and transparency: freeze audit + corpus overview.
2. Taiwan-side historical saturation: high binding, weak time movement, primary-source limitation.
3. Corridor comparison: overseas/regional records show where explicit Taiwan markers matter more.
4. Target controls: weak/no-binding records prevent the argument from becoming positive-only.
5. Authority sensitivity: retrospective tourism records are interpreted as memory discourse, not equal primary evidence.
6. Nostalgia/platform layer: old-flavor and street-snack markers connect Taiwaneseness to commodified memory.
7. Exception analysis: low-binding pockets and leave-one-source-type-out tests explain why an apparently obvious trend is still analytically useful.
